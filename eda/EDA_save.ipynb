```python
df_nn_eda = df_exa.copy()

df_nn_eda = df_nn_eda[(df_nn_eda['patient'] == '001')]
#df_nn_eda = df_nn_eda[(df_nn_eda['patient'] != '006')]
df_nn_eda = df_nn_eda[df_nn_eda['staining'] == '25']
df_nn_eda = df_nn_eda[df_nn_eda['Term'] != 'Tissue']

n_neighbours_list = [1, 2, 3, 5, 10]
patients = df_nn_eda['patient'].unique()
stainings = df_nn_eda['staining'].unique()

col_names = ['patient', 'Term']
new_col_names = ['count_healthy','count_sclerotic','count_dead', 'nn']
all_col_names = col_names
all_col_names.extend(new_col_names)

results = pd.DataFrame([], columns=new_col_names)

for patient in patients:
    for n_neighbors in n_neighbours_list:
        # Get the samples for the selected patient and staining
        df_subset = df_nn_eda[(df_nn_eda['patient'] == patient)].reset_index(drop=True).copy()

        # Fit the NN model to all samples
        neighbors_model = NearestNeighbors(n_neighbors=n_neighbors+1)
        neighbors_model.fit(df_subset[['Center X', 'Center Y']])
        
        # Get all neighbors
        distances, indices = neighbors_model.kneighbors(df_subset[['Center X', 'Center Y']])
        
        # Iterate over all samples
        for _ in range(len(df_subset)):
            
            count_dead = sum([1 for idx in indices[_][1:] if df_subset.iloc[idx]['Term'] == 'Dead'])
            count_sclerotic = sum([1 for idx in indices[_][1:] if df_subset.iloc[idx]['Term'] == 'Sclerotic'])
            count_healthy = sum([1 for idx in indices[_][1:] if df_subset.iloc[idx]['Term'] == 'Healthy'])
            
            df_subset.at[_, f'count_dead'] = count_dead
            df_subset.at[_, f'count_sclerotic'] = count_sclerotic
            df_subset.at[_, f'count_healthy'] = count_healthy
            df_subset.at[_, 'nn'] = n_neighbors
            

        results = pd.concat([results, df_subset[all_col_names]])

results_save = results
```

```python
def knn_graph_construction(X_geo, X, Y, k):
    neighbors_model = NearestNeighbors(n_neighbors=k+1)
    neighbors_model.fit(X_geo)
    
    # Get all neighbors
    distances, indices = neighbors_model.kneighbors(X_geo)

    adj_list = {i: neighbours[1:].tolist() for i, neighbours in enumerate(indices)}
    distances = {i: d[1:].tolist() for i, d in enumerate(distances)}
    Y = {i: y for i, y in enumerate(Y)}
    X_geo = {i: (x[1]['pos_x'], x[1]['pos_y']) for i, x in enumerate(X_geo.iterrows())}
    X = {i: x[1].values for i, x in enumerate(X.iterrows())}

    return adj_list, distances, Y, X_geo, X 

def radius_graph_construction(X_geo, X, Y, r):
    neighbors_model = NearestNeighbors(radius=r)
    neighbors_model.fit(X_geo)
    
    # Get all neighbors
    distances, indices = neighbors_model.kneighbors(X_geo)

    adj_list = {i: neighbours[1:].tolist() for i, neighbours in enumerate(indices)}
    Y = {i: y for i, y in enumerate(Y)}
    X_geo = {i: (x[1]['pos_x'], x[1]['pos_y']) for i, x in enumerate(X_geo.iterrows())}
    X = {i: x[1].values for i, x in enumerate(X.iterrows())}

    return adj_list, Y, X_geo, X 
```

```python
def edge_homophily(adj_list, Y):
    """
    Calculate the edge heterophily of a graph.
    
    Parameters:
    graph (dict): A dictionary where keys are nodes and values are lists of neighboring nodes.
    labels (dict): A dictionary where keys are nodes and values are the corresponding labels.
    
    Returns:
    float: The edge heterophily of the graph.
    """
    total_edges = 0
    heterophilic_edges = 0
    
    for node, neighbors in adj_list.items():
        for neighbor in neighbors:
            total_edges += 1
            if Y[node] == Y[neighbor]:
                heterophilic_edges += 1
    
    return heterophilic_edges / total_edges if total_edges > 0 else 0

def weighted_edge_homophily(adj_list, Y, distances):
    """
    Calculate the edge heterophily of a graph.
    
    Parameters:
    graph (dict): A dictionary where keys are nodes and values are lists of neighboring nodes.
    labels (dict): A dictionary where keys are nodes and values are the corresponding labels.
    
    Returns:
    float: The edge heterophily of the graph.
    """
    total_edges = 0
    total_edge_distance = 0
    homophile_edge_distance = 0
    
    for node, neighbors in adj_list.items():
        
        for i, neighbor in enumerate(neighbors):
            d = distances[node][i]
            total_edge_distance += d
            if Y[node] == Y[neighbor]:
                homophile_edge_distance += d

    return homophile_edge_distance / total_edge_distance if total_edge_distance > 0 else 0
                

def node_homophily(adj_list, Y):
    """
    Calculate the node homophily for each node in the graph.
    
    Parameters:
    graph (dict): A dictionary where keys are nodes and values are lists of neighboring nodes.
    labels (dict): A dictionary where keys are nodes and values are the corresponding labels.
    
    Returns:
    dict: A dictionary where keys are nodes and values are the homophily score for each node.
    """
    homophily_scores = {}
    
    for node, neighbors in adj_list.items():
        if len(neighbors) == 0:
            homophily_scores[node] = 0  # No neighbors means 0 homophily
            continue
        
        same_label_count = sum(1 for neighbor in neighbors if Y[neighbor] == Y[node])
        homophily_scores[node] = same_label_count / len(neighbors)
    
    return homophily_scores

def total_node_homophily(adj_list, Y):
    """
    Calculate the node homophily for the whole graph. Uses node_homophily().
    
    Parameters:
    graph (dict): A dictionary where keys are nodes and values are lists of neighboring nodes.
    labels (dict): A dictionary where keys are nodes and values are the corresponding labels.
    
    Returns:
    float: Total node homophily of the graph.
    """
    homophily_scores = node_homophily(adj_list, Y)
    sum_scores = sum([s for s in homophily_scores.values()])
    n = len(homophily_scores)

    return sum_scores / n

def class_homophily(adj_list, Y):
    """
    Calculate the class homophily for each class in Y.
    
    Parameters:
    graph (dict): A dictionary where keys are nodes and values are lists of neighboring nodes.
    labels (dict): A dictionary where keys are nodes and values are the corresponding labels.
    
    Returns:
    dict: A dictionary where keys are classes k and values are the homophily score for each class.
    """
    
    C = set(Y.values())
    N = len(Y)
    homophily_scores = {}
    
    for k in C:
        # Nodes in class
        nodes_k = [node for node, y in Y.items() if y == k]

        # Calculate h_k for this class
        homophile_neighbours_k = 0
        all_neighbours_k = 0

        for node in nodes_k:
            homophile_neighbours_k += sum([1 for u in adj_list[node] if Y[node] == Y[u]])
            all_neighbours_k += sum([1 for u in adj_list[node]])

        h_k = homophile_neighbours_k / all_neighbours_k

                
        # Number of nodes in class k
        d_v_k = len(nodes_k)

        # Calculate balanced homophily
        homophily_scores[k] = h_k - (d_v_k / N) 
    
    return homophily_scores

def total_class_homophily(adj_list, Y):
    """
    Calculate the class homophily for the whole graph. Uses class_homophily().
    
    Parameters:
    graph (dict): A dictionary where keys are nodes and values are lists of neighboring nodes.
    labels (dict): A dictionary where keys are nodes and values are the corresponding labels.
    
    Returns:
    float: Total class homophily of the graph.
    """    
    homophily_scores = class_homophily(adj_list, Y)
    sum_scores = sum([s for s in homophily_scores.values()])
    n_C = len(homophily_scores)

    return sum_scores / (n_C - 1)

```

```python
def generlised_edge_homophily(adj_list, X):
    """
    Calculate the generlised edge heterophily of a graph.
    
    Parameters:
    graph (dict): A dictionary where keys are nodes and values are lists of neighboring nodes.
    labels (dict): A dictionary where keys are nodes and values are the corresponding labels.
    
    Returns:
    float: The generalised edge heterophily of the graph.
    """
    total_edges = 0
    total_cosine = 0
    
    for node, neighbors in adj_list.items():
        for neighbor in neighbors:
            total_edges += 1
            total_cosine += cosine_similarity(X[node].reshape(1, -1), X[neighbor].reshape(1, -1))[0,0]
    
    return total_cosine / total_edges

def weighted_generalised_edge_homophily(adj_list, X, distances):
    """
    Calculate the weighted generlised edge heterophily of a graph. Here the cosine similarity is weighted by the distance 
    between the two nodes.
    
    Parameters:
    graph (dict): A dictionary where keys are nodes and values are lists of neighboring nodes.
    labels (dict): A dictionary where keys are nodes and values are the corresponding labels.
    
    Returns:
    float: The weighted generalised edge heterophily of the graph.
    """
    total_distance = 0
    total_cosine = 0
    
    for node, neighbors in adj_list.items():
        for i, (neighbor) in enumerate(neighbors):
            d = distances[node][i]
            total_distance += d
            total_cosine += cosine_similarity(X[node].reshape(1, -1), X[neighbor].reshape(1, -1))[0,0] * d
    
    return total_cosine / total_distance

def relative_cosine_edge_homophily(adj_list, X):
    """
    Calculate the generlised edge heterophily of a graph.
    
    Parameters:
    graph (dict): A dictionary where keys are nodes and values are lists of neighboring nodes.
    labels (dict): A dictionary where keys are nodes and values are the corresponding labels.
    
    Returns:
    float: The generalised edge heterophily of the graph.
    """
    generalised_cosine_sim = generlised_edge_homophily(adj_list, X)

    n_edges_full = 0
    full_connected_cosine_sim = 0
    for i, x1 in X.items():
        for j, x2 in X.items():
            if j!=i:
                full_connected_cosine_sim += cosine_similarity(x1.reshape(1,-1), x2.reshape(1,-1))
                n_edges_full +=1
    full_connected_cosine_sim /= n_edges_full 
    
    return generalised_cosine_sim / full_connected_cosine_sim 
```

```python
def selective_edge_homophily(adj_list, X, Y):
    """
    Calculate the edge heterophily of a graph.
    
    Parameters:
    graph (dict): A dictionary where keys are nodes and values are lists of neighboring nodes.
    labels (dict): A dictionary where keys are nodes and values are the corresponding labels.
    
    Returns:
    float: The edge heterophily of the graph.
    """
    total_edges = 0
    heterophilic_edges = 0
    
    for node, neighbors in adj_list.items():
        for neighbor in neighbors:
            total_edges += 1
            similarity = cosine_similarity(X[node].reshape(1, -1), X[neighbor].reshape(1, -1))[0,0]
            if ((similarity > 0) & (Y[node] == Y[neighbor])) or ((similarity <= 0) & (Y[node] != Y[neighbor])):
                heterophilic_edges += 1
    
    return heterophilic_edges / total_edges if total_edges > 0 else 0
```

```python
def neighborhood_heterophily(adj_list:dict, Y: dict):
    """
    Calculate the neighborhood heterophily measure for a graph.
    
    Parameters:
    adj_list (dict): A dictionary where keys are nodes and values are lists of neighboring nodes.
    Y (dict): A dictionary where keys are nodes and values are the corresponding class labels.
    num_classes (int): The number of unique classes in the graph.
    
    Returns:
    float: The neighborhood heterophily measure for the graph.
    """
    C = set(Y.values()) # Classes
    n_C = len(C)
    N = len(adj_list)  # Total number of nodes
    class_counts = {k: 0 for k in C}  # Count of nodes in each class

    # Step 1: Build the class-level neighborhood label distribution matrix
    class_neighbors = {k: [] for k in C}  # Dictionary to hold neighbors' label distributions
    
    for node, neighbors in adj_list.items():
        y_i = Y[node]
        class_counts[y_i] += 1
        
        # Initialize a zero array for the node's neighborhood label distribution
        neighbor_label_count = np.zeros(n_C)
        
        for neighbor in neighbors:
            y_j = Y[neighbor]
            neighbor_label_count[y_j] += 1
        
        # Normalize the distribution to create a row in A_k^N
        total_neighbors = len(neighbors)
        if total_neighbors > 0:
            neighbor_label_distribution = neighbor_label_count / total_neighbors
            class_neighbors[y_i].append(neighbor_label_distribution)
    
    # Step 2: Calculate the singular values for each class matrix A_k^N
    H_neighbor = 0
    for k in C:
        if class_counts[k] == 0:
            continue  # Skip if no nodes belong to class k
        
        A_k_N = np.array(class_neighbors[k])  # Class-level neighborhood label distribution matrix
        
        # Perform Singular Value Decomposition (SVD) on the matrix
        if A_k_N.size > 0:
            U, S, Vt = np.linalg.svd(A_k_N, full_matrices=False)
            singular_values = S / np.sum(S)  # Normalize singular values
        
            # Step 3: Calculate entropy H_neighbor^k using normalized singular values
            entropy = -np.sum(singular_values * np.log(singular_values + 1e-9)) / np.log(n_C)
            
            # Weight by the proportion of nodes in class k
            H_neighbor += (class_counts[k] / N) * entropy
    
    return H_neighbor
```

```python
def cosine_similarity_custom(vec1, vec2):
    vec1 = np.array(vec1)
    vec2 = np.array(vec2)
    
    dot_product = np.dot(vec1, vec2)
    magnitude_vec1 = np.linalg.norm(vec1)
    magnitude_vec2 = np.linalg.norm(vec2)
    
    if magnitude_vec1 == 0 or magnitude_vec2 == 0:
        return 0
    return dot_product / (magnitude_vec1 * magnitude_vec2)
```

```python
n_neighbours_list = [1, 2, 3, 5, 10]
patients = df_exc['patient'].unique()
stainings = df_exc['staining'].unique()

col_names = ['patient', 'staining', 'Term']
new_col_names = [f'dead n={n}' for n in n_neighbours_list]
new_col_names.extend([f'sclerotic n={n}' for n in n_neighbours_list])
all_col_names = col_names
all_col_names.extend(new_col_names)

results = pd.DataFrame([], columns=new_col_names)

for patient in patients:
    for staining in stainings:
        # Get the samples for the selected patient and staining
        df_subset = df_exc[(df_exc['patient'] == patient) & (df_exc['staining'] == staining)]

        # Get all unhealthy gloms from this patient in this staining
        df_unhealthy = df_subset[(df_subset['Term'] == 'Dead') | (df_subset['Term'] == 'Sclerotic')].reset_index(drop=True)

        for col in new_col_names:
            df_unhealthy[col] = 0 

        for n_neighbors in n_neighbours_list:

            # Fit the NN model to all samples
            neighbors_model = NearestNeighbors(n_neighbors=n_neighbors+1)
            neighbors_model.fit(df_subset[['Center X', 'Center Y']])
            
            # Get all neighbors
            distances, indices = neighbors_model.kneighbors(df_unhealthy[['Center X', 'Center Y']])
            
            # Iterate over all unhealthy samples
            for _ in range(len(df_unhealthy)):
                
                count_dead = sum([1 for idx in indices[_][1:] if df_subset.iloc[idx]['Term'] == 'Dead'])
                count_sclerotic = sum([1 for idx in indices[_][1:] if df_subset.iloc[idx]['Term'] == 'Sclerotic'])
                
                df_unhealthy.at[_, f'dead n={n_neighbors}'] = count_dead
                df_unhealthy.at[_, f'sclerotic n={n_neighbors}'] = count_sclerotic

        results = pd.concat([results, df_unhealthy[all_col_names]])
            

    
```
