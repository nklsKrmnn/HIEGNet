```python
stain_list = [25,106,107,108]
list_annotation_paths = [f'/home/dascim/data/1_cytomine_downloads/EXC/annotations/{s}/annotations_001_{s}.csv' for s in stain_list]
list_annotation_df = [pd.read_csv(path) for path in list_annotation_paths]
```

```python
for i, df_ann in enumerate(list_annotation_df):
    list_annotation_df[i]=df_ann[df_ann['Term'] != 'Tissue']
```

```python
list_annotation_df[0].info()
```

```python
for stain, df_stain in zip(stain_list, list_annotation_df):
    id_col = f'{stain}_id'
    term_col = f'Term_{stain}'
    
    df = df.merge(df_stain[['ID', 'Term']], left_on=id_col, right_on='ID', how='outer')
    df.rename(columns={'Term': term_col}, inplace=True)
    df.drop(columns=['ID'], inplace=True)
```

```python
# List of columns to check
term_columns = [f'Term_{stain}' for stain in stain_list]

# Define a function to categorize each row
def categorize_row(row):
    if any(row[col] == 'Dead' for col in term_columns):
        return 'Dead'
    elif any(row[col] == 'Sclerotic' for col in term_columns):
        return 'Sclerotic'
    else:
        return 'Healthy'

# Apply the function to each row
df['Category'] = df.apply(categorize_row, axis=1)
```

```python
pre_drop = df['Category'].value_counts()
post_drop = df.dropna()['Category'].value_counts()
diff = pre_drop - post_drop
diff_rel = diff / pre_drop
```
